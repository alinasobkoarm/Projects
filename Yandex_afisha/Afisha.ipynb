<b>Комментарий студента:</b> Ключ event_id проверен. Дубликатов по этому ключу в таблице с событиями нет. Данные можно объединять. Те 19 event_id, которые есть в таблице orders, но которых нет в таблице events, не попадут в общую таблицу.
</div>

````python
# исключаем days_since_prev при проверке
order_check_cols = ['user_id', 'created_dt_msk', 'created_ts_msk', 'event_id', 
                    'cinema_circuit', 'age_limit', 'currency_code', 'device_type_canonical', 
                    'revenue', 'service_name', 'tickets_count', 'total']

# Находим неявные дубликаты в orders
duplicates = orders.duplicated(subset=order_check_cols, keep=False).sum()
print(duplicates)
````

````python
#удаляем дубликаты
orders_clean = orders.drop_duplicates(subset=order_check_cols, keep='first').copy()
print(len(orders) - len(orders_clean))
````

````python
#пересчитаем дни между заказами после удаления дубликатов 
# преобразуем created_ts_msk в datetime
orders_clean['created_ts_msk'] = pd.to_datetime(orders_clean['created_ts_msk'])
orders_clean = orders_clean.sort_values(['user_id', 'created_ts_msk'])
orders_clean['days_since_prev'] = orders_clean.groupby('user_id')['created_ts_msk'].diff().dt.days
````

````python
#находим неявные дубликаты в событиях 
duplicated_1 = events[events.duplicated(subset=['event_id'], keep=False)]
print(len(duplicated_1))
````

````python
# колонки для проверки содержания (все, кроме идентификатора)
content_cols = ['event_name', 'event_type_description', 'event_type_main', 
                'organizers', 'region_name', 'city_name', 'venue_name', 'venue_address']
# находим дубликаты по содержанию 
duplicates_2 = events[events.duplicated(subset=content_cols, keep=False)]
display(duplicates_2.sort_values(['event_name', 'event_id']).head(10))
````

````python
#удаляем дубликаты по содержанию. В таблице events могут существовать разные event_id, но с одинаковым содержанием 
#(название, описание, организаторы, регион, город, площадка, адрес).
events_clean = events.drop_duplicates(subset=content_cols, keep='first')
print(len(events_clean))
````

````python
#считаем, сколько удалено
print(len(events)-len(events_clean))
````

````python
#объединим датасеты orders и events в один
df = orders_clean.merge(events_clean, on='event_id')
display(df)
````

При интерпретации речь, разумеется, тоже должна идти о заказах. В свою очередь **общее среднее** подразумевает - раз мы считаем среднюю выручку со стоимости билета, то и работать мы должны с билетами - в числителе берем суммарную выручку, в знаменателе берем суммарное количество купленных билетов. В этом случае мы уже оцениваем величину выручки от продажи **среднестатистического билета**. Результаты по этим двум подходам будут отличаться. Причем, чем разнороднее наши заказы по объему билетов и затрат, тем существеннее будут разниться результаты этих двух подходов, поскольку при расчете среднего от средних мы уравниваем вклад каждого заказа, влияние экстремальных заказов с большим количеством билетов на показатель снижается (в общем среднем они продавливают показатели своей кучей билетов), а влияние пользователей с заказов количеством билетов, напротив, нарастает (в общем среднем они теряются среди остальных, задавливаются).

    Но, несмотря на то, что среднее от средних в целом можно еще как-то проинтерпретировать, по факту его практически не используют и, говоря про среднее, подразумевают общее среднее, оно гораздо понятнее. Через сколь много уровней мы бы его не протаскивали и какими бы сильными не были бы различия в объемах групп, это будет стабильная в интерпретации величина - среднее на 1 билет. В свою очередь среднее от средних при неравенстве объема групп может давать искажение уже на первом шаге, а если мы начинаем его использовать многократно (сначала при расчете заказных средних, потом на основе них для общезаказного показателя по сегментам / дням, потом еще по по более верхнеуровневым сегментам), то оно станет вовсе неинтерпретируемым, основание у такого среднего многоуровневое, что довольно сложно осмыслить, возникает много "если бы".

Объем выборок превышает 30 наблюдений, что позволяет применять параметрические критерии благодаря Центральной предельной теореме.<br>
2.Единица наблюдения
Единицей наблюдения является пользователь. Для каждого пользователя считается общее количество заказов, сделанных с мобильного и десктопного устройства.<br>
3.Независимость наблюдений и выборок<br>
20.5% пользователей используют оба типа устройств, что создает зависимость между выборками. Для учета этого эффекта пользователи разделены на три группы:
Для независимых групп используется ttest_ind.
Для зависимой группы (оба устройства) используется ttest_rel (парный тест), так как один и тот же пользователь представлен в обеих выборках.<br>
4.Сравниваемый параметр<br>
Сравнивается среднее арифметическое количество заказов на пользователя.<br>
5.Объем выборки<br>
Размер выборок превышает 30, что делает применение t-теста статистически обоснованным.<br>
6.Выбросы и асимметрия<br>
Проверка показала наличие асимметрии и выбросов. Однако большой объем выборки делает t-тест работоспособным к нарушениям нормальности.

Гипотеза: среднее количество заказов на одного пользователя мобильного приложения выше по сравнению с пользователями стационарных устройств.

Нулевая гипотеза H₀: среднее количество заказов на одного пользователя мобильного приложения такое же, как у пользователей стационарных устройств

Альтернативная гипотеза H₁: среднее количество заказов на одного пользователя мобильного приложения выше по сравнению с пользователями стационарных устройств, и это различие статистически значимо.

````python
# разделяем пользователей на 3 группы
mobile_users = set(autumn[autumn['device_type_canonical'] == 'mobile']['user_id'].unique())
desktop_users = set(autumn[autumn['device_type_canonical'] == 'desktop']['user_id'].unique())
both_users = mobile_users & desktop_users
only_mobile_users = mobile_users - desktop_users
only_desktop_users = desktop_users - mobile_users

# считаем заказы на пользователя
user_orders = autumn.groupby(['user_id', 'device_type_canonical'])['order_id'].count().reset_index()
user_orders.columns = ['user_id', 'device_type_canonical', 'orders_per_user']
only_mobile_orders = user_orders[user_orders['user_id'].isin(only_mobile_users)]['orders_per_user']
only_desktop_orders = user_orders[user_orders['user_id'].isin(only_desktop_users)]['orders_per_user']
both_mobile = user_orders[(user_orders['user_id'].isin(both_users)) & (user_orders['device_type_canonical'] == 'mobile')]['orders_per_user']
both_desktop = user_orders[(user_orders['user_id'].isin(both_users)) & (user_orders['device_type_canonical'] == 'desktop')]['orders_per_user']

# проводим тест для пользующихся только мобильными или только стационарными устройствами
alpha = 0.05
t_stat1, p_value1 = ttest_ind(only_mobile_orders, only_desktop_orders, alternative='greater')
if p_value1 < alpha:
    print("Нулевая гипотеза отвергнута. Пользователи только мобильных устройств совершают статистически значимо больше заказов, чем пользователи только десктопных устройств.")
else:
    print("Нет оснований отвергать нулевую гипотезу. Статистически значимых различий между пользователями только мобильных и только десктопных устройств не обнаружено.")

# парный тест для пользователей с обоими устройствами
t_stat2, p_value2 = ttest_rel(both_mobile, both_desktop, alternative='greater')
if p_value2 < alpha:
    print("Нулевая гипотеза отвергнута. У пользователей, использующих оба устройства, заказы с мобильного статистически значимо превышают заказы с десктопа.")
else:
    print("Нет оснований отвергать нулевую гипотезу. У пользователей с обоими устройствами статистически значимых различий между мобильными и десктопными заказами не обнаружено.")
````

    Для начала разберем саму проблему. В первой гипотезе единицей наблюдения является пользователь, совершающий заказы. Однако один и тот же пользователь может оформлять заказы как с мобильного устройства, так и с десктопа. При этом у каждого клиента есть собственный уровень активности и лояльности, влияющий на общее количество заказов. Из-за этого заказы одного и того же пользователя с разных устройств оказываются статистически зависимыми и перестают быть независимыми наблюдениями. Если пользователь совершил больше заказов с одного устройства, то количество заказов с другого устройства уже частично определяется его общей активностью. Такое нарушение независимости может искажать результаты тестирования и усложнять интерпретацию эффекта.

    Самый простой способ решить проблему - оставить только пользователей, использующих одно устройство. Однако в таком случае мы теряем значительную часть аудитории и ничего не узнаем о пользователях со смешанным паттерном поведения, хотя их может быть даже больше, чем пользователей только десктопной версии. Поэтому здесь важно сначала определить, что именно мы хотим тестировать.

    1. Один из вариантов - разделить пользователей на три группы:
 
        * пользователи мобильных устройств
        * пользователи десктопных устройств
        * пользователи, использующие оба типа устройств

      После этого можно попарно сравнить группы между собой. Такой подход позволит понять, в какой группе находятся более активные или лояльные пользователи, однако смешанная группа все равно останется довольно неоднородной.

    <br/>
        
    2. Чтобы уменьшить эту неоднородность, можно дополнительно разделить смешанных пользователей по предпочитаемому устройству - например, на пользователей с преимущественно мобильным или преимущественно десктопным паттерном заказов.
 
    <br/>
        
    3. Можно подойти к задаче и иначе, немного изменив исследовательский вопрос. Например, не сравнивать пользователей между собой, а проверить, существует ли систематический перекос в количестве заказов между мобильными и десктопными устройствами. В таком случае для каждого пользователя нужно посчитать:
 
        * количество заказов с мобильного устройства
        * количество заказов с десктопного устройства

      Важно, чтобы в обеих выборках присутствовали все пользователи, включая тех, у кого по одному из устройств количество заказов равно нулю. После этого можно рассмотреть разность между ними и проверить, отличается ли средняя разность от нуля. Такой подход естественным образом приводит к парному тестированию, поскольку сравнение проводится внутри одного и того же пользователя. Для этого подходит парный `t-test` - [ttest_rel](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_rel.html), который сводит задачу к анализу распределения попарных разностей и проверяет, отличается ли их среднее от нуля. Поскольку внутри используется критерий Стьюдента, важно анализировать именно распределение разностей, а не исходных выборок. Если распределение разностей не содержит тяжелых хвостов, сильных выбросов и имеет достаточный объем наблюдений, то `ttest_rel` отлично подходит. Если же предположения критерия Стьюдента выполняются плохо, альтернативой выступает [непараметрический критерий Вилкоксона](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.wilcoxon.html).
        
    <br/>

    </details>

    <details> <summary> <span style='color:blue' > Вторая гипотеза (тык) </span> </summary>

    Со второй гипотезой ситуация чуть сложнее. Если взять пользователя как единицу наблюдения и отдельно усреднить интервалы между его мобильными и десктопными заказами, то останется незафиксированным сам порядок смены устройств. Пользователь может чередовать устройства совершенно по-разному:

    * где-то использовать только одно устройство
    * где-то часто переключаться между ними
    * где-то возвращаться к одному и тому же устройству сериями

  Из-за этого часть эффекта смешивается, и интерпретация становится неоднозначной. Если же взять единицей наблюдения отдельные заказы и просто разделить их по текущему устройству, то остается неучтенным влияние предыдущего устройства. Предыдущий пользовательский опыт может быть связан с дальнейшим поведением клиента, и тогда мы снова получаем смешение эффектов. Поэтому здесь особенно важно правильно определить единицу наблюдения. В данном случае естественной единицей наблюдения является не пользователь и не отдельный заказ, а интервал между двумя соседними заказами. Такой интервал характеризуется как минимум двумя признаками:

    * устройством предыдущего заказа
    * устройством текущего заказа

  Здесь тоже можно использовать несколько подходов.

    1. Первый вариант - аналогично первой гипотезе разделить пользователей на группы:

        * преимущественно мобильные
        * преимущественно десктопные
        * пользователи со смешанным паттерном использования устройств

      После этого можно посчитать средний интервал между заказами для каждого пользователя и сравнить полученные пользовательские средние между группами. Такой подход уже даст достаточно информативную картину различий в пользовательском поведении. Однако можно пойти дальше.
        
    <br/>

    2. Можно проанализировать непосредственно переходы между устройствами. Для этого можно сформировать четыре группы переходов:

        * desktop → desktop
        * desktop → mobile
        * mobile → desktop
        * mobile → mobile

      После этого сравнивать интервалы между заказами уже между конкретными типами переходов. Например:

        * сравнить переходы без смены устройства и оценить, какое устройство связано с более быстрым возвращением пользователя
        * рассмотреть только переходы со сменой устройства и проверить, после какого устройства пользователи чаще меняют платформу
        * зафиксировать предыдущее устройство и проверить, связано ли оно с возвратом на то же самое или другое устройство
        * либо, наоборот, зафиксировать текущее устройство и анализировать, с какого устройства пользователи чаще приходят

    <br/>

    3. Дополнительно можно сегментировать пользователей по устройству первого заказа и анализировать, как стартовый пользовательский опыт связан с дальнейшей структурой заказов и пользовательской лояльностью.

    </details>


* <details> <summary> <span style='color:blue' > Нюансы интерпретации (тык) </span> </summary>
    
    * В задании сейчас говорится про средние показатели, при этом объем данных у нас достаточно большой, поэтому использование критерия Стьюдента на самом деле вполне оправдано. Но важно понимать, что, ввиду присутствия сильной асимметрии, он раскрывает не все нюансы поведения клиентов, поэтому имеет смысл также смотреть результаты непараметрического теста, например `Mann–Whitney U test` (формулировка гипотез, соответственно, тоже меняется, в этом случае мы уже сравниваем не средние значения). Данный критерий работает с рангами наблюдений, а не с самими значениями, поэтому он менее чувствителен к выбросам. А при схожей форме распределений его результаты можно интерпретировать как различия в медианах.

        Интересный момент - если результаты параметрического и непараметрического тестов противоречат друг другу, это может указывать на различия в характере распределений: влияние выбросов и хвостов, высокую дисперсию или различие формы распределений между группами. В таком случае распределения стоит дополнительно изучить визуально и через описательные статистики.

        В данном проекте, как правило (конечно, зависит еще от предобработки), в отношении интервалов между заказами образуется противоречие результатов. В нашем случае большую роль играют хвосты и нули (вероятно, перекупы, совершающие множественные серии небольших заказов). Поэтому данному моменту имеет смысл уделить дополнительное внимание.

    <br/>
    
    * В плане интерпретации есть еще важный момент. Гипотезы - это не "состояние мира", которое мы можем подтвердить или не подтвердить. `H0` это математическая модель данных, распределение статистики при условии выполнения нулевой гипотезы. В свою очередь `H1` это множество альтернатив. При этом сам тест непосредственно работает именно с нулевой гипотезей, а не альтернативной, поэтому суждения мы можем строить лишь в отношении `H0`, а отношение к `H1` уже является следствием из этого. 

        Разберем подробнее, как мы работаем с нулевой гипотезой (в классическом случае - отсутствия различий или эффекта). Относительно этой модели строится критерий отклонения: мы проверяем, насколько наблюдаемые данные совместимы с нулевой гипотезой, контролируя вероятность ложного отвержения нулевой гипотезы, но не оценивая вероятность отсутствия эффекта (вероятность верности нулевой гипотезы). Поэтому между формулировками "подтвердить `H0`" и "не отвергнуть `H0`" существует принципиальная разница: во втором случае речь идет лишь о недостатке доказательств против нулевой гипотезы, что может быть связано с малым размером выборки, низкой мощностью теста, высокой дисперсией, шумными данными или плохим дизайном эксперимента.

    <br/>
    
    * Отмечу также, что характеристики выборки и условия применимости статистических критериев важно учитывать не только на этапе выбора теста, но и при интерпретации результатов. Размер выборки, вариабельность данных (дисперсия) и величина эффекта напрямую влияют на статистическую мощность - вероятность обнаружения реальных различий при их наличии.

        Кроме того, важно различать статистическую и практическую значимость результатов. При больших объемах данных даже очень небольшие различия могут становиться статистически значимыми, но не иметь практической ценности. В то же время при малых выборках содержательно важный эффект может не достигать статистической значимости из-за недостаточной мощности теста.

        Учет этих факторов позволяет сделать интерпретацию результатов более полной и методологически корректной, а также избежать формального сводения анализа к p-value без оценки реальной величины эффекта.

  </details>

    * **Подавление системных предупреждений (Warnings)**
      * **Что делать:** Перед сдачей работы исправлять причины возникновения `Warning`. Если это предупреждение библиотеки, которое нельзя обойти, временно заглушать его через `warnings.filterwarnings('ignore')`.
      * **Почему:** Полотна розовых предупреждений от Python между кодом и графиками выглядят грязно и создают ощущение "сырого" решения.
    
    <br/>
    
    * **Опрятность логов циклов и прогресс-баров**
      * **Что делать:** При обучении моделей или парсинге не выводить тысячи строк итераций. Использовать библиотеку `tqdm` для красивой одной строки прогресса.
      * **Почему:** Бесконечный скролл логов убивает читаемость проекта.

5. <details> <summary> <span style='color:blue' > Визуализация (тык) </span> </summary>
    
    * **"Один график - одна мысль" и минимализм**
      * **Что делать:** Не пытаться уместить на один чарт 10 распределений. Использовать цвет, маркеры и текст только ради смысла, а не "для красоты". Если нужно что-то сравнить - реализуем это на одном графике, а не на двух соседних, так как сравнение - и есть основная задача.
      * **Почему:** Избыточные элементы перегружают восприятие и маскируют главный инсайт.
    
    <br/>
    
    * **Самодостаточность графика**
      * **Что делать:** Обязательно добавлять заголовки и подписи (`title`, `xlabel`, `ylabel`) и легенду (если на графике больше одного цвета / маркера). Подбирать оптимальный размер (`figsize`).
      * **Почему:** График должен быть понятен, даже если его скопируют в презентацию или рабочий чат в отрыве от кода и текста работы.
    
    <br/>
    
    * **Честность масштаба и фокус на данных (Ось Y)**
      * **Что делать:** Для столбчатых диаграмм (`Bar-charts`) и линейных графиков объемов (выручка, пользователи) **всегда начинать ось Y с нуля**. Для диаграмм рассеяния (`Scatter-plots`) и графиков относительных метрик (конверсия, маржинальность, точность моделей) **настраивать масштаб под диапазон данных**, не привязываясь к нулю.
      * **Почему:** В первом случае обрезка оси искусственно раздувает мелкие колебания и пугает бизнес. Во втором случае привязка к нулю "сплющит" график, лишит его детализации и скроет от аналитика реальные тренды и инсайты.
    
    <br/>
    
    * **Читаемость текста и чистый вывод (`plt.show()`)**
      * **Что делать:** Завершать код графика вызовом `plt.show()` (или `;`). Если подписи осей наслаиваются - разворачивать их (`rotation=45`) или менять график на горизонтальный.
      * **Почему:** Убирает мусорные строки технического вывода библиотеки (`<Axes: ...>`) и предотвращает наложение текста.
    
    <br/>
    
    * **Единая палитра для сущностей (смысловой цвет)**
      * **Что делать:** Если на первом графике категории покрашены в определенные цвета (например, "Отток" - красный, "Активные" - синий), эта кодировка должна сохраняться на всех графиках проекта.
      * **Почему:** Мозг читателя привыкает к паттерну. Смена цветов в середине работы запутает аудиторию.
    
    <br/>
    
    * **Сортировка категорий в Bar-charts**
      * **Что делать:** Всегда сортировать столбчатые диаграммы по убыванию или возрастанию целевого значения (кроме естественного порядка, вроде дней недели или месяцев).
      * **Почему:** Забор из хаотичных по высоте столбиков невозможно анализировать глазами. Сортировка сразу подсвечивает лидеров и аутсайдеров. Наличие логического порядка позволяет проследить динамику.
    
    <br/>
    
    * **Правильный выбор типа графика**
      * **Что делать:** Строго соблюдать правила:
        * Не использовать `Pie-chart` (круговую диаграмму), если категорий больше 3–4. Как вариант - малые категории объединить в одну.
        * Не использовать `Line-plot` (линейный график) для несвязанных категориальных данных (только для непрерывных величин или временных рядов).
      * **Почему:** Нарушение этих правил физически искажает восприятие данных человеческим глазом.
    
    <br/>
    
    * **Цветовая доступность и отказ от дефолтных палитр**
      * **Что делать:** Избегать одновременного использования чистого красного и зеленого цветов для контрастных данных (проблема дальтонизма). Использовать современные сбалансированные палитры (например, из библиотеки `seaborn`).
      * **Почему:** Дефолтные кислотные цвета Matplotlib выглядят непрофессионально, а неправильный контраст делает график нечитаемым для части аудитории.

6. <details> <summary> <span style='color:blue' > Культура содержательного анализа (от фактов к инсайтам) (тык) </span> </summary>
    
    * **Постановка задач "от себя"**
      * **Что делать:** Переводить общие требования брифа в плоскость конкретных подзадач анализа внутри ноутбука.
      * **Почему:** Это демонстрирует самостоятельность мышления, а не слепое следование инструкциям.
    
    <br/> 

    * **Запрет на пересказ визуализации (отказ от роли "капитана очевидность")**
      * **Что делать:** Никогда не писать в выводах то, что и так видно глазами. Вместо фразы *"ы видим, что в декабре пик продаж, а в январе - спад"* переходить на уровень причин и бизнес-контекста.
      * **Почему:** Пересказ графиков тратит время читателя и не несет новой информации. Аналитика ценна не констатацией факта, а ответом на вопрос **"Почему так произошло?"**.

    <br/>
    
    * **Правило трех вопросов: "Что? → Почему? → И что с того?"**
      * **Что делать:** Каждое важное наблюдение в данных проводить по трехэтапному фильтру аналитического мышления:
        1. **Что произошло?** (Факт: *"В группе клиентов от 18 до 23 лет отток составляет 40%"*).
        2. **Почему так могло получиться?** (Гипотезы / причины: *"У этой аудитории низкий средний чек, они чувствительны к ценам конкурентов, либо наш интерфейс для них слишком сложный"*).
        3. **И что с того? / Какая польза бизнесу?** (Действие: *"Это критический инсайт. Нам нужно либо запустить для них студенческий тариф, чтобы удержать, либо перестать тратить маркетинг на их привлечение, так как они не окупаются"*).
      * **Почему:** Только третий шаг ("И что с того?") превращает сухие цифры в ценность, за которую бизнес готов платить деньги.

    <br/>
    
    * **Фильтрация инсайтов: "интересно vs бесполезно"**
      * **Что делать:** Оценивать свои находки с точки зрения полезности для принятия решений. 
        * *Бесполезный инсайт:* Очевидная сезонность (например, *"летом продажи мороженого выше"*). Об этом бизнес знает и без аналитика.
        * *Интересный (контринтуитивный) инсайт:* Аномалия или слом шаблона (например, *"мы думали, что скидки привлекают лояльных клиентов, но данные показывают, что они лишь увеличивают отток после окончания акции"*).
      * **Почему:** Если перегрузить итоговый отчет банальными фактами, важные и по-настоящему глубокие инсайты просто потеряются в этом шуме.

    <br/>
    
    * **Поиск причинно-следственных связей (корреляция != причинность)**
      * **Что делать:** Прежде чем делать громкие заявления, проверять альтернативные объяснения. Если продажи выросли одновременно с запуском новой рекламной кампании, нужно проверить, не совпало ли это с праздниками, выплатой зарплат или уходом крупного конкурента.
      * **Почему:** Ложные выводы приводят к неверным бизнес-решениям и потере бюджетов. Хороший аналитик всегда выступает в роли скептика и проверяет скрытые факторы.
    
    <br/>
    
    </details>
