Усовершенствуем нашу функцию

````python
# перепишем функцию!

def sum_of_two(first_value, second_value):
    """
    Функция, которая выводит на принт сумму двух чисел

    Параметры:
        first_value - первое слагаемое, число
        second_value - второе слагаемое, число

    Вернет:
        Принтит сумму двух чисел
        Возвращает None
    """
    return first_value + second_value
````

#### Цикл for

Цикл for в Python обычно используется для итерации по элементам последовательности (например, списка, кортежа, словаря, множества или строки). Этот тип цикла идеален, когда заранее известно количество итераций


````python
fruits = ["яблоко", "банан", "груша"]

for fruit in fruits:
    print(fruit)
````

#### Цикл while

Цикл while выполняется, пока условие остается истинным. Этот тип цикла предпочтителен, когда количество итераций неизвестно заранее

````python
count = 1
while count <= 5:
    print(count)
    count += 1  # увеличиваем счетчик на 1 после каждой итерации
    
````

#### Ключевое слово break

`break` используется для немедленного выхода из цикла, прерывая его дальнейшее выполнение, даже если условие цикла ещё не стало ложным

````python
for number in range(1, 10):
    if number == 5:
        break  # Прерываем цикл, если число достигло 5
    print(number)
````

#### Ключевое слово continue

`continue` пропускает оставшуюся часть кода в текущей итерации и немедленно переходит к следующей итерации цикла

````python
for number in range(1, 10):
    if number == 5:
        continue  # Прерываем цикл, если число достигло 5. Продолжаем цикл дальше
    print(number)
````

````python
# Что выведет код? Почему?
for number in range(1, 10):
    if number % 2 == 0:
        continue  # Пропускаем четные числа
    print(number)
````

## Обработка ошибок try - except

### Исключения и ошибки в Python

> Add blockquote



Исключения — это способ, которым Python сообщает о том, что что-то пошло не так в программе. Они могут возникать по разным причинам, например, из-за деления на ноль, обращения к несуществующему ключу в словаре или попытки открыть файл, который не существует.

### try и except примеры

````python
x = 5
y = 0
result = x / y
    
print('Что-то важное')
````

````python
try:
    x = 5
    y = 0
    result = x / y
except:
    print("Ошибка: Деление на ноль!")
    
print('Что-то важное')
````

````python
# Обработка ошибки деления на ноль
try:
    x = 5
    y = 0
    result = x / y
except:
    pass

print('Код выполнился до конца')
````

## Зоны видимости в Python

### Локальная область видимости

Локальная область видимости относится к переменным, определенным внутри функции. Эти переменные доступны только внутри функции и не могут быть доступны из других частей программы, включая другие функции, если только они не передаются как аргументы. Обратите внимание на эту задачу - она часто попадается на собеседованиях.


````python
def my_function():
    local_var = 10  # Локальная переменная, определенная внутри функции
    print(local_var)  # Доступ к переменной внутри функции

local_var = 12
print(local_var)
my_function()
print(local_var)
````

### Глобальная область видимости

Глобальные переменные определяются в основной части программы и могут быть доступны из любой части кода, включая функции. Чтобы изменить глобальную переменную внутри функции, необходимо использовать ключевое слово `global`

````python
global_var = 5
abcd = 28
# Глобальная переменная

def get_global():    
    print(abcd)
    
get_global()
````

````python
global_var
````

````python
global_var = 5  # Глобальная переменная

def modify_global():
    global global_var
    global_var = 20
     # Изменение глобальной переменной внутри функции
````

````python
print("До вызова функции:", global_var)
````

````python
modify_global()
````

````python
print("После вызова функции:", global_var)
````

## Понятие чистый код


# Основные правила PEP 8

### 1. Отступы
- Используй **4 пробела** для одного уровня отступа (никаких табов!).

### 2. Длина строки
- Не более **79 символов** (72 для комментариев).

### 3. Пробелы
- Не ставь пробелы перед запятыми, внутри скобок, или вокруг `=` в аргументах функций.  
  ```python
  func(arg1=2, arg2=5)  # правильно
  ```

### 4. Именование
- **Переменные и функции**: `snake_case`.  
- **Классы**: `CamelCase`.  
- **Константы**: `UPPER_CASE`.

### 5. Импорты
- Ставь импорты в начале файла.
- Сначала **стандартные библиотеки**, потом **сторонние**, затем **локальные модули**.


### 6. Комментарии
- Должны быть **краткими**, **полезными**, обновляться вместе с кодом.

### 7. Читаемость
- Код должен быть **понятным** и **простым**.
