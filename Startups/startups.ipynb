### 2.1. Группы по срокам финансирования

Разделите все компании на три группы:

* Единичное финансирование — был всего один раунд финансирования.

* Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.

* Срок финансирования более года.

Визуализируйте соотношение этих групп, создав два графика:

* По количеству компаний: Покажите, какой процент от общего числа компаний относится к каждой из трёх групп.
* По объёму инвестиций: Отобразите, какую долю от общего объёма привлечённых средств получила каждая группа.

Совет: Для ясности и согласованности используйте единую цветовую палитру для всех графиков, чтобы каждая категория (например, «Единичное финансирование») всегда отображалась одним цветом.

```python
#считаем количество раундов
rounds = ['round_a', 'round_b', 'round_c', 'round_d', 'round_e', 'round_f', 'round_g', 'round_h']
total_rounds = (df[rounds] > 0).sum(axis=1)
#cчитаем разницу между первым и последним раундом финансирования в днях
days_diff = (df['last_funding_at'] - df['first_funding_at']).dt.days
#создан новый столбец со сроком финансирования
df['funding_group'] = 'Срок финансирования более года'  
df.loc[total_rounds <= 1, 'funding_group'] = 'Единичное финансирование'
df.loc[(total_rounds > 1) & (days_diff <= 365), 'funding_group'] = 'Срок финансирования до года'
#создаем сводную таблицу
groups = df.groupby('funding_group').agg(количество_компаний=('funding_group', 'count'), объем_инвестиций=('funding_total_usd', 'sum')
).reset_index()
#добавляем столбец с объемом инвестиций в миллионах долларов
groups['объем_инвестиций'] = (groups['объем_инвестиций'] / 1_000_000).round(2)
#добавляем данные для визуализаций
total_companies = groups['количество_компаний'].sum()
total_investment = groups['объем_инвестиций'].sum()
groups['доля_компаний, %'] = (groups['количество_компаний'] / total_companies * 100).round(2)
groups['доля_инвестиций'] = (groups['объем_инвестиций'] / total_investment).round(2)

print(groups.to_string(index=False))
```

```python
colors = ['royalblue', 'magenta', 'orange']
groups['доля_компаний, %'].plot(kind='bar', color=colors,figsize=(10, 6),legend=False)
plt.title('Доля компаний в зависимости от срока финансирования')
plt.ylabel('Доля компаний, %')
plt.xlabel('Срок финансирования: 0-единичное финансирование, 1 - срок финансирования более года, 2 - срок финансирования до года')
plt.xticks (rotation=0)
plt.grid(axis='y', alpha=0.3, linewidth=0.5, color='black')
plt.show()
```

Определите компании с аномальным объёмом общего финансирования — используйте метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определите сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведите топ таких сегментов.

```python
markets = df['market'].unique()
df['anomaly'] = False
```

```python
for market_type in markets:
    market_data = df[df['market'] == market_type]
    q1 = market_data['funding_total_usd'].quantile(0.25)
    q3 = market_data['funding_total_usd'].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper =  q3 + 1.5 * iqr
    df.loc[ (df['market'] == market_type) &
            ( (df['funding_total_usd'] < lower) |
            (df['funding_total_usd'] > upper ) )
           ,'anomaly'] = True
```

### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам. Ваша цель — для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Совет: когда будете делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавьте к знаменателю небольшое число, например `1e-60`. Это поможет избежать деления на ноль.

Постройте график, на котором отобразите нормированные значения возврата средств для типов финансирования `venture`, `debt_financing`, `private_equity`, `seed` и `angel`.

Сделайте вывод о том, в каких типах финансирования наблюдается наиболее устойчивый рост показателя.

```python
import numpy as np
sums_by_year = (df.groupby('year')[finance_type].sum())/1_000_000
sums_by_year = sums_by_year.sort_index(ascending=True)
returns_by_year = df2.groupby('year')[finance_type].sum()
returns_by_year = returns_by_year.sort_index(ascending=True)
display(sums_by_year)
display(returns_by_year)
```

```python
epsilon = 1e-1
ratios_by_year = returns_by_year / (sums_by_year + epsilon)
# Убираем аномалии 
ratios_by_year_clean = ratios_by_year.copy()
# Удаление аномалий по каждому году отдельно
for year in ratios_by_year.index:
    year_data = ratios_by_year.loc[year]
    year_data_clean = year_data.dropna()
    if len(year_data_clean) > 0:       
        Q1 = year_data_clean.quantile(0.25)
        Q3 = year_data_clean.quantile(0.75)
        IQR = Q3 - Q1        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        for col in ratios_by_year.columns:
            value = ratios_by_year.loc[year, col]
            if not pd.isna(value):
                if value < lower_bound or value > upper_bound:
                    ratios_by_year_clean.loc[year, col] = np.nan
print(ratios_by_year_clean.round(4))
print(ratios_by_year.round(4))
```

```python
selected_types = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']
ratios_selected = ratios_by_year_clean[selected_types].copy()
plt.figure(figsize=(24, 12))
for col in ratios_selected.columns:
    plt.plot(ratios_selected.index, ratios_selected[col],
         marker='o', linewidth=2, markersize=8,label=col )
plt.title('Коэффициенты возвратов')
plt.xlabel('Год')
plt.ylabel('Коэффициенты возвратов')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5), fontsize=12) 
plt.grid(True, alpha=0.3)
plt.xticks(ratios_selected.index, rotation=45)
```

Роста возвратов среди выбранных категорий не наблюдается.
